```python
df["Review Rating"]= df.groupby("Category")["Review Rating"].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ","_")
df =df.rename(columns={"purchase_amount_(usd)":"purchase_amount"})
```

```python
df.columns
```

```python
#create a column 
labels = ["Young adult","adult","Middle-age","Senior"]
df["age_group"] = pd.qcut(df["age"],q=4, labels= labels)
```

```python
df[["age","age_group"]].head(10)
```

```python
#create a column purchase_frequency_days

frequency_mapping = {
    "Fortnightly": 14,
    "Weekly":7,
    "Monthly":30,
    "Quarterly":90,
    "Bi-Weekly":14,
    "Annually":365,
    "Every 3 Months":90
}

df["purchase_frequency_days"]=df["frequency_of_purchases"].map(frequency_mapping)
```

```python
import pandas as pd
from sqlalchemy import create_engine
from sqlalchemy.engine import URL
import os

username = os.environ.get("CUSTOMER_DB_USER", "root")
password = os.environ["CUSTOMER_DB_PASSWORD"]
host = os.environ.get("CUSTOMER_DB_HOST", "127.0.0.1")
port = int(os.environ.get("CUSTOMER_DB_PORT", "3306"))
database = os.environ.get("CUSTOMER_DB_NAME", "mydatabase")

connection_url = URL.create(
    drivername="mysql+pymysql",
    username=username,
    password=password,
    host=host,
    port=port,
    database=database
)

engine = create_engine(connection_url)

table_name = "mytable"

df.to_sql(
    name=table_name,
    con=engine,
    if_exists="replace",
    index=False
)

sample = pd.read_sql(
    f"SELECT * FROM {table_name} LIMIT 5;",
    con=engine
)

print(sample)
```
